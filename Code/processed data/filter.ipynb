```python
import pandas as pd

# Path to your CSV
csv_path = "analyzed_agricultural_data.csv"

# Crops of interest (adjust names if needed to match your data)
crops_of_interest = [
    "Cereals, primary",
    "Maize (corn)",
    "Vegetables primary",
    "Rice"
]

# Load CSV
df = pd.read_csv(csv_path)

# Inspect possible column naming variants (optional, for robustness)
# You can print(df.columns) to confirm exact header names
# Expected columns: Area, Item, Year, Value
# If headers differ, adjust these mappings accordingly.

# Normalize column names (robust to case and extra spaces)
df.columns = [c.strip() for c in df.columns]
df.columns = [c.lower() for c in df.columns]

# Ensure required columns exist (case-insensitive handling)
required = ["area", "item", "year", "value"]
missing = [r for r in required if r not in df.columns]
if missing:
    raise ValueError(f"Missing required columns in CSV: {missing}")

# Filter for China and crops of interest
filtered = df[
    (df["area"] == "China") &
    (df["item"].isin(crops_of_interest))
]

# Optional: convert Year to int and Value to float (in case they're strings)
filtered["year"] = filtered["year"].astype(int)
filtered["value"] = pd.to_numeric(filtered["value"], errors="coerce")

# Sort for readability
filtered = filtered.sort_values(by=["year", "item"])

# Output: print to console
print(filtered)

# Optional: save to a new CSV for downstream use
output_path = "filtered_china_crops.csv"
filtered.to_csv(output_path, index=False)
print(f"\nFiltered data saved to: {output_path}")
```
