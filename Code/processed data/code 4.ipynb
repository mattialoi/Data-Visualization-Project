```python
import pandas as pd

# Load your data
df = pd.read_csv('nutrients.csv')

# First, melt to long format to get fertilizer types as values
df_long = df.melt(
    id_vars=['Entity', 'Code', 'Year'],
    var_name='Fertilizer_Type',
    value_name='Tonnes'
)

# Clean up the fertilizer type names
df_long['Fertilizer_Type'] = df_long['Fertilizer_Type'].str.replace('Nutrient ', '')

# Now pivot to get years as columns
df_pivot = df_long.pivot_table(
    index=['Entity', 'Code', 'Fertilizer_Type'],
    columns='Year',
    values='Tonnes'
).reset_index()

# Sort years in ascending order
sorted_years = sorted([col for col in df_pivot.columns if isinstance(col, int)])
other_cols = ['Entity', 'Code', 'Fertilizer_Type']
df_pivot = df_pivot[other_cols + sorted_years]

print("Final DataFrame preview:")
print(df_pivot.head())
print(f"\nShape: {df_pivot.shape}")
print(f"Fertilizer Types: {df_pivot['Fertilizer_Type'].unique().tolist()}")
print(f"Years as columns: {sorted_years[:5]}... {sorted_years[-5:]}")

# Save the results
df_pivot.to_csv('nutrients_fertilizer_years.csv', index=False)
print("\nSaved to 'nutrients_fertilizer_years.csv'")
```
