```python
import pandas as pd

# 1. Carica il DataFrame (sostituisci 'your_data.csv' con il nome del tuo file)
df = pd.read_csv('world_population.csv')

# --- Preparazione: Identifica le colonne ---

# 2. Definisci le colonne da trasformare (gli anni)
# Converti i numeri interi (anni) in stringhe, se non lo sono già nel tuo CSV.
year_cols = [str(year) for year in range(1961, 2024)] # Crea la lista: ['1961', '1962', ..., '2025']

# 3. Identifica le colonne che NON devono essere trasformate (ID columns)
# Ad esempio, se hai una colonna 'Entity', la metterai qui.
# Se non hai altre colonne oltre agli anni, usa id_vars=[]
id_cols = ['Country Name', 'Country Code', 'Indicator Name', 'Indicator Code' ] # Sostituisci con i nomi delle tue colonne non-anno

# 4. Applica la trasformazione "melt" (Unpivot)
df_long = df.melt(
    id_vars=id_cols,              # Colonne da mantenere fisse (ID)
    value_vars=year_cols,         # Colonne da trasformare in righe (gli anni)
    var_name='Year',              # Nome della nuova colonna che conterrà i nomi delle vecchie colonne (gli anni)
    value_name='Value'            # Nome della nuova colonna che conterrà i valori
)

# 5. Convergi la colonna 'Year' da stringa a intero (opzionale, ma consigliato)
df_long['Year'] = df_long['Year'].astype(int)

# 6. Salva il nuovo DataFrame in formato lungo (Long Format)
output_csv_filename = 'world_data_elab.csv'
df_long.to_csv(output_csv_filename, index=False)

print(f"La trasformazione è stata completata. Il file in formato lungo è salvato come: {output_csv_filename}")
```
